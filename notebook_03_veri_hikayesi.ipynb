## 1. Veri Setinin Kapsamı ve Amaç

Bu çalışma Hamamözü, Gümüşhacıköy ve Göynücek ilçelerine ait elektrik tahakkuk ve tahsilat verileri üzerinden gerçekleştirilmiştir.

Amaç; ilçeler, hesap sınıfları ve zaman boyutu üzerinden tüketim ve ödeme davranışlarını analiz ederek,
karar vericilere yol gösterecek öngörüler elde etmektir.


## 2. İlçeler Arası Karşılaştırma

Gümüşhacıköy ilçesi, benzersiz müşteri sayısı açısından diğer ilçelere kıyasla belirgin biçimde öne çıkmaktadır.
Bu durum, ilçenin hem nüfus hem de ekonomik faaliyet yoğunluğunun daha yüksek olduğunu göstermektedir.

Hamamözü ise daha sınırlı bir müşteri profiline sahip olup,
tüketim ve tahsilat analizlerinde daha düşük hacimli bir ilçe olarak konumlanmaktadır.


## 3. Hesap Sınıflarına Göre Tüketim Davranışı

Birçok hesap sınıfında ortalama tüketim değerlerinin medyanın oldukça üzerinde olduğu görülmektedir.
Bu durum, yüksek tüketimli az sayıda abonenin dağılımı yukarı çektiğini göstermektedir.

Özellikle sanayi, arıtma tesisleri ve lisanssız üreticiler gibi gruplarda
yüksek standart sapma değerleri, tüketimin homojen olmadığını ortaya koymaktadır.


## 4. Mevsimsel Tüketim Analizi

Aylık ortalama tüketim değerleri incelendiğinde,
özellikle yaz aylarında (Haziran–Ağustos) tüm ilçelerde belirgin bir tüketim artışı gözlemlenmektedir.

Hamamözü ilçesinin diğer ilçelere kıyasla daha stabil bir tüketim profiline sahip olması,
ilçedeki sanayi ve yüksek hacimli abonelerin sınırlı olduğunu düşündürmektedir.


## 5. Tahsilat ve Ödeme Davranışı

Tahsilat verileri incelendiğinde,
tahakkukların %66.1’inin son ödeme tarihinden önce veya zamanında ödendiği,
%33.9’luk kısmın ise gecikmeli olarak tahsil edildiği görülmektedir.

Bu oran, önemli bir müşteri kitlesinin ödeme alışkanlıklarının iyileştirilebileceğine işaret etmektedir.


## 6. Tüketim Dağılımı ve Aykırı Değerler

kWh tüketim dağılımı histogram ve box plot ile incelendiğinde,
0 kWh tüketim değerlerinin çok düşük frekansa sahip olduğu görülmektedir.

Box plot analizine göre tüketimlerin büyük bölümü 0–50.000 kWh aralığında yoğunlaşırken,
50.000 kWh üzerindeki değerler aykırı gözlemler olarak öne çıkmaktadır.
Bu durum, belirli abonelerin sistem üzerinde orantısız etkiye sahip olduğunu göstermektedir.


## 7. Sonuç ve Öneriler

Analiz sonuçları, ilçeler ve hesap sınıfları arasında belirgin tüketim ve ödeme davranışı farklılıkları olduğunu göstermektedir.

• Yüksek tüketimli ve yüksek varyanslı hesap sınıfları için özel izleme ve raporlama mekanizmaları kurulabilir.  
• Geç ödeme oranını azaltmak amacıyla hedefli bilgilendirme ve teşvik politikaları uygulanabilir.  
• Mevsimsel tüketim artışları dikkate alınarak kapasite ve tahsilat planlamaları yapılabilir.
