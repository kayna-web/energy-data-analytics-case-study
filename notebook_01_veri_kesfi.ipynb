## kWh Değişkeninde Veri Kalitesi Kontrolü

Bu bölümde birleşik tahakkuk veri setinde yer alan
kwh değişkeni özelinde veri kalitesi kontrolü yapılmıştır.

Amaç, analiz sonuçlarını yanıltabilecek eksik değerleri,
negatif tüketimleri ve aşırı uç gözlemleri tespit ederek
ilerleyen adımlarda nasıl ele alınmaları gerektiğine
karar vermektir.


```python
df_tahakkuk_all["kwh"].isna().sum()
```

```python
(df_tahakkuk_all["kwh"] < 0).sum()
```

```python
df_tahakkuk_all["kwh"].describe()
```

```python
Q1 = df_tahakkuk_all["kwh"].quantile(0.25)
Q3 = df_tahakkuk_all["kwh"].quantile(0.75)
IQR = Q3 - Q1
alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR
outlier_sayisi = (
    (df_tahakkuk_all["kwh"] < alt_sinir) |
    (df_tahakkuk_all["kwh"] > ust_sinir)
).sum()
alt_sinir, ust_sinir, outlier_sayisi
```

## Hesap Sınıfına Göre Elektrik Tüketimlerinin İncelenmesi

Bu bölümde birleşik tahakkuk veri seti kullanılarak
hesap sınıflarına göre elektrik tüketim istatistikleri hesaplanmıştır.

Amaç, farklı abone türlerinin (hesap sınıflarının)
ortalama tüketim düzeylerini, tipik tüketim davranışlarını
(medyan) ve tüketim değişkenliğini
(standart sapma) karşılaştırmalı olarak incelemektir.


```python
hesap_sinifi_istatistikleri = (
 df_tahakkuk_all
  .groupby("Hesap Sınıfı")["kwh"]
   .agg(
     ortalama_kwh="mean",
     medyan_kwh="median",
     std_kwh="std").reset_index())
hesap_sinifi_istatistikleri
```

### Hesap Sınıfına Göre Elektrik Tüketimi Bulguları

Hesap sınıfına göre hesaplanan ortalama, medyan ve standart sapma
değerleri incelendiğinde, elektrik tüketim davranışlarının
abonelik türlerine göre önemli ölçüde farklılaştığı görülmektedir.

- En yüksek ortalama ve medyan tüketim değerleri,
  **Karayolları Genel Müdürlüğü Aydınlatma**,
  **Arıtma Tesisleri** ve **Lisansız Üreticiler** hesap sınıflarında
  gözlemlenmiştir. Bu gruplar, yüksek ve sürekli enerji ihtiyacı olan
  altyapı ve üretim faaliyetlerini yansıtmaktadır.
- **Sanayi** ve **Tarımsal Faaliyetler** gibi üretim odaklı hesap
  sınıflarında ortalama tüketim değerleri yüksek olmakla birlikte,
  medyan değerlerin ortalamaya göre daha düşük olması,
  tüketimin sınırlı sayıdaki yüksek hacimli aboneler tarafından
  yukarı çekildiğini göstermektedir.
- Mesken hesap sınıfında ortalama (55,59 kWh) ve medyan
  (47,36 kWh) değerlerinin birbirine yakın olması,
  konut abonelerinde daha homojen bir tüketim yapısına işaret etmektedir.
- Kamu, belediye ve resmi kurumlara ait hesap sınıflarında
  standart sapma değerlerinin yüksek olması,
  bu gruplar içinde tüketim düzeylerinin oldukça değişken
  olduğunu göstermektedir.
- İbadethane, bina ortak kullanım ve benzeri düşük tüketimli
  hesap sınıflarında ortalama ve medyan değerlerin düşük,
  standart sapmanın sınırlı olduğu görülmektedir.

Genel olarak, standart sapma değerlerinin birçok hesap sınıfında
ortalama tüketim değerlerine kıyasla yüksek olması,
kWh değişkeninde uç değer etkisinin devam ettiğini göstermektedir.
